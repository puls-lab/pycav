```python
%matplotlib widget
import matplotlib.pyplot as plt
import numpy as np
import pycav
```

```python
cwl = 1.03e-6
aoi = np.deg2rad(5)
```

### standard example
**Note:** the variation plots can be clicked on

```python
cav = pycav.Cavity(1e-6)
cav.add_mirror(0, 0, idn="SESAM")
cav.add_space(1, idn="S_TD")
cav.add_mirror(1, aoi, idn="TD")
cav.add_space(0.2, idn="TD_R1")
cav.add_mirror(2, -aoi, idn="R1")
cav.add_space(0.4, idn="R1_OC")
cav.add_mirror(0, 0, idn="OC")

varpar = [
    ("S_TD", "d", np.linspace(0, 3, 100)),
    ("TD_R1", "d", np.linspace(0, 3, 100)),
    ("SESAM", "r", np.linspace(0, 3, 100)),
    ("TD", "r", np.linspace(0.1, 10, 100)),
    ("TD", "aoi", np.linspace(0, np.deg2rad(45), 46)),
]

elsofin = ["TD", "SESAM", "OC"]

cdc = pycav.CavityDesigner(cav, varpar, elsofin)
```

### example with equal-ID sections

```python
cav = pycav.Cavity(1e-6)
cav.add_mirror(2, 0, idn="TD")
cav.add_space(1, idn="L1")
cav.add_mirror(2, aoi, idn="TD")
cav.add_space(1, idn="L1")
cav.add_mirror(2, -aoi, idn="R1")
cav.add_space(0.4, idn="R1_OC")
cav.add_mirror(0, 0, idn="OC")

varpar = [("L1", "d", np.linspace(0, 3, 100)), ("TD", "r", np.linspace(0.1, 10, 100))]

elsofin = ["TD", "OC"]

cdc = pycav.CavityDesigner(cav, varpar, elsofin)
```

# beam propagation

```python
cav = pycav.Beam(1e-6)
cav.set_w0([1e-3, 1e-3])
# cav.add_mirror(0, 0, 'SESAM')
cav.add_space(1, idn="a")
cav.add_lens([1, 1], idn="L1")
cav.add_space(1, idn="b")
cav.add_lens([1, 1], idn="L2")
cav.add_space(1, idn="c")
cav.add_lens([1, 1], idn="L3")
cav.add_space(1, idn="d")

varpar = [
    ("a", "d", np.linspace(0, 3, 101)),
    ("L1", "f1", np.linspace(0, 10, 101)),
    ("L2", "fs", np.linspace(0, 10, 101)),
    ("L3", "ft", np.linspace(0, 10, 101)),
]

elsofin = ["L1", "L2", "L3"]

cdb = pycav.gui.CavityDesigner(cav, varpar, elsofin)
```
