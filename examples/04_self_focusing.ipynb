## Self-focusing in a bow-tie cavity

A Kerr medium at the focus between the two curved mirrors of a bow-tie ring
cavity acts as an intensity-dependent lens. This example shows how the cavity
mode changes with the intracavity peak power. Each colored band spans the
sagittal and tangential beam radius for one power; lower powers are drawn on
top, so the self-focusing shows up as the colored fringes.

```python
%matplotlib widget
import matplotlib.pyplot as plt
import numpy as np
import pycav
```

```python
def bowtie(P):
    """Bow-tie ring cavity with a 3 mm Kerr medium at the focus.

    P is the intracavity peak power in W."""
    aoi = np.deg2rad(8)
    cav = pycav.Cavity(1030e-9, step=1e-4, ring=True)
    cav.add_mirror(r=0.2, aoi=aoi, idn="CM1")
    cav.add_space(d=0.103)
    cav.add_kerr_medium(n=1.76, n2=3e-20, d=3e-3, P=P, idn="KM")
    cav.add_space(d=0.103)
    cav.add_mirror(r=0.2, aoi=aoi, idn="CM2")
    cav.add_space(d=0.5)
    cav.add_mirror(r=0, aoi=aoi, idn="FM1")
    cav.add_space(d=0.2)
    cav.add_mirror(r=0, aoi=aoi, idn="FM2")
    cav.add_space(d=0.5)
    return cav
```

```python
powers = np.linspace(0, 2e6, 100)
cmap = plt.cm.rainbow
norm = plt.Normalize(powers.min() / 1e6, powers.max() / 1e6)

fig, ax = plt.subplots(figsize=(9, 3))
# highest power first, so that lower powers are drawn on top
for P in powers[::-1]:
    d, w = bowtie(P).get_beam()
    ax.fill_between(d, w[:, 0] * 1e3, w[:, 1] * 1e3, color=cmap(norm(P / 1e6)))
fig.colorbar(
    plt.cm.ScalarMappable(norm=norm, cmap=cmap), ax=ax, label="peak power / MW"
)
ax.set_xlabel("position in cavity / m")
ax.set_ylabel("beam radius / mm")
fig.tight_layout()
```

With increasing power, the Kerr lens shrinks the focus in the Kerr medium,
and the beam in the long arm grows, more strongly in the sagittal plane. The
Kerr lens is calculated iteratively (`kerrit`, default 25 steps), which is
converged here.
